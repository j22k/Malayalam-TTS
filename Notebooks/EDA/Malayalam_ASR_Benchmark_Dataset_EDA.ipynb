```python
import zipfile
import json

ZIP_PATH = "./downloads/aikosh_dataset/Kathbath-Malayalam-Test-Known.zip"

with zipfile.ZipFile(ZIP_PATH, "r") as z:
    with z.open("Kathbath-Malayalam-Test-Known/data.json") as f:
        data = json.load(f)

print("Python type:", type(data))

if isinstance(data, list):
    print("Number of records:", len(data))

    print("\nFirst record:")
    print(json.dumps(data[0], ensure_ascii=False, indent=2))

elif isinstance(data, dict):
    print("Top-level keys:")
    print(list(data.keys()))

    print("\nPreview:")
    print(json.dumps(data, ensure_ascii=False, indent=2)[:3000])
```

```python
from collections import Counter

records = data

# Basic counts
print("Total records:", len(records))

# Required fields
required_fields = ["audioFilename", "text", "gender", "speaker"]

print("\nMissing values:")
for field in required_fields:
    missing = sum(
        1 for r in records
        if field not in r or r[field] is None or str(r[field]).strip() == ""
    )
    print(f"  {field}: {missing}")

# Speaker distribution
speakers = Counter(r.get("speaker") for r in records)

print("\nNumber of unique speakers:", len(speakers))

print("\nSpeaker distribution:")
for speaker, count in speakers.most_common():
    print(f"  Speaker {speaker}: {count}")

# Gender distribution
genders = Counter(r.get("gender") for r in records)

print("\nGender distribution:")
for gender, count in genders.items():
    print(f"  {gender}: {count}")

# Duplicate audio references
audio_files = [r.get("audioFilename") for r in records]
audio_counts = Counter(audio_files)

duplicates = {
    filename: count
    for filename, count in audio_counts.items()
    if count > 1
}

print("\nDuplicate audio references:", len(duplicates))

if duplicates:
    print("\nDuplicate examples:")
    for filename, count in list(duplicates.items())[:10]:
        print(f"  {filename}: {count} times")

# Duplicate transcripts
texts = [r.get("text", "").strip() for r in records]
text_counts = Counter(texts)

duplicate_texts = {
    text: count
    for text, count in text_counts.items()
    if count > 1
}

print("\nDuplicate transcript texts:", len(duplicate_texts))
```

```python
import zipfile
import json
import io
import wave
from collections import Counter

ZIP_PATH = "./downloads/aikosh_dataset/Kathbath-Malayalam-Test-Known.zip"

with zipfile.ZipFile(ZIP_PATH, "r") as z:

    # Get all actual files in the audio directory
    audio_files = {
        name.replace("Kathbath-Malayalam-Test-Known/", "")
        for name in z.namelist()
        if name.startswith("Kathbath-Malayalam-Test-Known/audios/")
        and name.lower().endswith(".wav")
    }

    print("WAV files in ZIP:", len(audio_files))

    sample_rates = Counter()
    channels = Counter()
    sample_widths = Counter()
    durations = []

    for record in records:
        relative_path = record["audioFilename"]

        # Expected ZIP path
        zip_path = "Kathbath-Malayalam-Test-Known/" + relative_path

        if zip_path not in z.namelist():
            continue

        with z.open(zip_path) as f:
            audio_bytes = f.read()

        with wave.open(io.BytesIO(audio_bytes), "rb") as wav:
            sr = wav.getframerate()
            ch = wav.getnchannels()
            width = wav.getsampwidth()
            frames = wav.getnframes()

            duration = frames / sr

            sample_rates[sr] += 1
            channels[ch] += 1
            sample_widths[width] += 1
            durations.append(duration)

    print("\nSample rates:")
    for sr, count in sample_rates.items():
        print(f"  {sr} Hz: {count}")

    print("\nChannels:")
    for ch, count in channels.items():
        print(f"  {ch} channel(s): {count}")

    print("\nSample widths:")
    for width, count in sample_widths.items():
        print(f"  {width * 8}-bit: {count}")

    print("\nDuration:")
    print(f"  Minimum: {min(durations):.2f} sec")
    print(f"  Maximum: {max(durations):.2f} sec")
    print(f"  Average: {sum(durations) / len(durations):.2f} sec")
    print(f"  Total: {sum(durations) / 3600:.2f} hours")
```

```python
import re
import statistics

texts = [r["text"].strip() for r in records]

lengths_chars = [len(t) for t in texts]
lengths_words = [len(t.split()) for t in texts]

print("Transcript statistics")
print("-" * 40)

print(f"Characters:")
print(f"  Minimum : {min(lengths_chars)}")
print(f"  Maximum : {max(lengths_chars)}")
print(f"  Average : {statistics.mean(lengths_chars):.1f}")

print(f"\nWords:")
print(f"  Minimum : {min(lengths_words)}")
print(f"  Maximum : {max(lengths_words)}")
print(f"  Average : {statistics.mean(lengths_words):.1f}")

# Detect characters that are not Malayalam letters,
# Malayalam combining marks, whitespace, or common punctuation.
unexpected = Counter()

for text in texts:
    for ch in text:
        if not (
            "\u0D00" <= ch <= "\u0D7F"   # Malayalam
            or ch.isspace()
            or ch in ".,!?;:'\"()-–—%0123456789"
        ):
            unexpected[ch] += 1

print("\nUnexpected character types:")
print("  Number of unique characters:", len(unexpected))

for char, count in unexpected.most_common(30):
    print(f"  {repr(char)} : {count}")

# Show shortest and longest examples
shortest_idx = lengths_chars.index(min(lengths_chars))
longest_idx = lengths_chars.index(max(lengths_chars))

print("\nShortest transcript:")
print(texts[shortest_idx])

print("\nLongest transcript:")
print(texts[longest_idx])
```

```python
from collections import defaultdict
import zipfile
import io
import wave

ZIP_PATH = "./downloads/aikosh_dataset/Kathbath-Malayalam-Test-Known.zip"

speaker_stats = defaultdict(list)

with zipfile.ZipFile(ZIP_PATH, "r") as z:

    for record in records:
        zip_path = "Kathbath-Malayalam-Test-Known/" + record["audioFilename"]

        audio_bytes = z.read(zip_path)

        with wave.open(io.BytesIO(audio_bytes), "rb") as wav:
            duration = wav.getnframes() / wav.getframerate()

        speaker_stats[record["speaker"]].append(duration)

print("Speaker statistics")
print("-" * 60)

for speaker, durations in sorted(
    speaker_stats.items(),
    key=lambda x: sum(x[1]),
    reverse=True
):
    print(
        f"Speaker {speaker:>4} | "
        f"Gender: {next(r['gender'] for r in records if r['speaker'] == speaker):<6} | "
        f"Files: {len(durations):>3} | "
        f"Hours: {sum(durations)/3600:.2f} | "
        f"Avg: {sum(durations)/len(durations):.2f}s"
    )
```

```python
import re

working_records = [
    r for r in records
    if r["speaker"] != "117"
]

# Malayalam Unicode block
malayalam_pattern = re.compile(r"[\u0D00-\u0D7F]")

stats = {
    "punctuation": 0,
    "digits": 0,
    "english": 0,
    "other": 0,
}

examples = {
    "punctuation": [],
    "digits": [],
    "english": [],
    "other": [],
}

for record in working_records:
    text = record["text"]

    for char in text:
        if char.isdigit():
            stats["digits"] += 1
            if len(examples["digits"]) < 5:
                examples["digits"].append(text)

        elif char.isascii() and char.isalpha():
            stats["english"] += 1
            if len(examples["english"]) < 5:
                examples["english"].append(text)

        elif not malayalam_pattern.match(char) and not char.isspace():
            stats["punctuation"] += 1
            if len(examples["punctuation"]) < 5:
                examples["punctuation"].append(text)

print("Transcript character check")
print("-" * 60)

for key, value in stats.items():
    print(f"{key:12}: {value}")

print("\nExamples:")
for key, values in examples.items():
    if values:
        print(f"\n{key}:")
        for text in values:
            print(" ", text)
```

```python
from collections import Counter

speaker_counts = Counter(
    record["speaker"]
    for record in working_records
)

print("Usable speaker distribution")
print("-" * 50)

for speaker, count in sorted(
    speaker_counts.items(),
    key=lambda x: x[1],
    reverse=True
):
    print(f"Speaker {speaker:>4}: {count:>3} files")

print("\nTotal usable files:", sum(speaker_counts.values()))
print("Total usable speakers:", len(speaker_counts))
```
