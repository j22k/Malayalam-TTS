```python
# ============================================================
# CELL 9 — Test F5-TTS
# ============================================================

from huggingface_hub import hf_hub_download
from hydra.utils import get_class
from omegaconf import OmegaConf

from f5_tts.infer.utils_infer import (
    infer_process,
    load_model,
    load_vocoder,
    preprocess_ref_audio_text,
)

print("F5-TTS imports successful")
```

```python
# ============================================================
# CELL 10 — Download F5-TTS OpenBible Malayalam files
# ============================================================
#
# We download:
#   model_last.pt  -> trained Malayalam F5-TTS checkpoint
#   vocab.txt      -> model vocabulary
#   YAML config    -> model architecture/configuration
# ============================================================

from huggingface_hub import hf_hub_download

REPO_ID = "multilingual-tts/F5-TTS-OpenBible-Malayalam"

ckpt = hf_hub_download(
    repo_id=REPO_ID,
    filename="model_last.pt"
)

vocab = hf_hub_download(
    repo_id=REPO_ID,
    filename="vocab.txt"
)

config = hf_hub_download(
    repo_id=REPO_ID,
    filename="F5-TTS_OpenBible_Malayalam.yaml"
)

print("Checkpoint:", ckpt)
print("Vocabulary:", vocab)
print("Config     :", config)
```

```python
# ============================================================
# CELL 11 — Load Malayalam model configuration
# ============================================================

from omegaconf import OmegaConf

model_cfg = OmegaConf.load(config)

print(model_cfg)
```

```python
# ============================================================
# CELL 12 — Select inference device
# ============================================================

import torch

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Inference device:", device)

if device == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
# ============================================================
# CELL 13 — Get F5-TTS model class
# ============================================================

from hydra.utils import get_class

model_cls = get_class(
    f"f5_tts.model.{model_cfg.model.backbone}"
)

print("Model class:", model_cls)
```

```python
# ============================================================
# CELL 14 — Load Vocos vocoder
# ============================================================
#
# The Malayalam model uses Vocos and generates audio at 24 kHz.
# ============================================================

from f5_tts.infer.utils_infer import load_vocoder

vocoder = load_vocoder(
    vocoder_name="vocos",
    is_local=False,
    device=device,
)

print("Vocos loaded successfully")
```

```python
# ============================================================
# CELL 15 — Load Malayalam F5-TTS checkpoint
# ============================================================

from f5_tts.infer.utils_infer import load_model

model = load_model(
    model_cls,
    model_cfg.model.arch,
    ckpt,
    mel_spec_type="vocos",
    vocab_file=vocab,
    use_ema=True,
    device=device,
)

print("Malayalam F5-TTS model loaded successfully")
```

```python
# ============================================================
# CELL 16 — Record reference voice from Colab microphone
# ============================================================
#
# Record roughly 5–10 seconds.
#
# Speak naturally in Malayalam.
# Keep:
#   - quiet environment
#   - one speaker
#   - no music/background speech
#   - normal speaking speed
#
# IMPORTANT:
# We need the EXACT text you speak because F5-TTS uses the
# reference transcript together with the reference audio.
# ============================================================

from IPython.display import Javascript, display
from google.colab import output
import base64

display(Javascript(r"""
async function recordAudio() {
    const stream = await navigator.mediaDevices.getUserMedia({audio: true});

    const recorder = new MediaRecorder(stream);
    const chunks = [];

    recorder.ondataavailable = event => {
        if (event.data.size > 0) {
            chunks.push(event.data);
        }
    };

    recorder.start();

    alert("Press OK, then speak Malayalam for about 5–10 seconds.");

    await new Promise(resolve => {
        recorder.onstop = resolve;
        setTimeout(() => recorder.stop(), 10000);
    });

    stream.getTracks().forEach(track => track.stop());

    const blob = new Blob(chunks, {type: "audio/webm"});
    const reader = new FileReader();

    return new Promise(resolve => {
        reader.onloadend = () => {
            resolve(reader.result);
        };
        reader.readAsDataURL(blob);
    });
}
"""))

audio_data = output.eval_js("recordAudio()")

audio_bytes = base64.b64decode(
    audio_data.split(",")[1]
)

with open("/content/reference_raw.webm", "wb") as f:
    f.write(audio_bytes)

print("Saved: /content/reference_raw.webm")
```

```python
# ============================================================
# CELL 17 — Convert reference recording to WAV
# ============================================================
#
# F5-TTS works with WAV input.
#
# We convert:
#   WebM → WAV
#
# and resample to 24 kHz.
# ============================================================

from pydub import AudioSegment

audio = AudioSegment.from_file(
    "/content/reference_raw.webm"
)

audio = audio.set_channels(1)
audio = audio.set_frame_rate(24000)

audio.export(
    "/content/reference.wav",
    format="wav"
)

print("Saved: /content/reference.wav")
print("Duration:", round(len(audio) / 1000, 2), "seconds")
```

```python
# ============================================================
# CELL 18 — Listen to reference audio
# ============================================================

from IPython.display import Audio, display

display(
    Audio(
        "/content/reference.wav",
        autoplay=False
    )
)
```

```python
# ============================================================
# CELL 19 — Reference transcript
# ============================================================
#
# IMPORTANT:
# This MUST match what you actually said in the recording.
# Don't paraphrase it.
# ============================================================

ref_text = """
എന്റെ പേര് ജുനൈദ് ആണ്. ഞാൻ കേരളത്തിൽ നിന്നാണ്.
""".strip()

print("Reference text:")
print(ref_text)
```

```python
# ============================================================
# CELL 20 — Target Malayalam text
# ============================================================

gen_text = """
ഇന്ന് കേരളത്തിൽ നല്ല കാലാവസ്ഥയാണ്. ഞാൻ മലയാളത്തിൽ സ്വാഭാവികമായി സംസാരിക്കുന്നു.
""".strip()

print("Target text:")
print(gen_text)
```

```python
# ============================================================
# CELL 21 — Preprocess reference audio + transcript
# ============================================================

from f5_tts.infer.utils_infer import preprocess_ref_audio_text

ref_audio_proc, ref_text_proc = preprocess_ref_audio_text(
    "/content/reference.wav",
    ref_text
)

print("Reference preprocessing complete")
print("Processed audio:", ref_audio_proc)
print("Processed text :", ref_text_proc)
```

```python
# ============================================================
# CELL 22 — ZERO-SHOT VOICE CLONING
# ============================================================
#
# Input:
#   reference voice
#   reference transcript
#   new Malayalam text
#
# Output:
#   generated Malayalam speech in the reference voice
# ============================================================

from f5_tts.infer.utils_infer import infer_process

wav, sr, _ = infer_process(
    ref_audio_proc,
    ref_text_proc,
    gen_text,
    model,
    vocoder,
    mel_spec_type="vocos",
    device=device,
)

print("Generation complete")
print("Sample rate:", sr)
print("Samples:", len(wav))
```

```python
# ============================================================
# CELL 23 — Save generated speech
# ============================================================

import soundfile as sf

output_path = "/content/malayalam_zero_shot.wav"

sf.write(
    output_path,
    wav,
    sr
)

print("Saved:", output_path)
```
